In this exercise, we want to look at the performance of different approaches to common string problems. For each problem, Experiment with the string length and add loops to run the code repeatedly for more reliable results. First, we need to generate some DNA sequences of arbitrary length.

```python
import random
import time
def generate_random_strings(n_strings: int, len_strings: int) -> list:
    chars = ["A", "T", "C", "G"]
    return [''.join(random.choices(chars, k=len_strings)) for _ in range(n_strings)]
```

I wrote an additional function to show the time taken. I called it benchmark and use it in the following problems several times.
The code is below. I also created a completed .py file which will be included in the final submission.
I used code I previously wrote during my EPI codereview for a similar time test program.

```python
def benchmark(name: str, func, *args, iterations: int = 1000):
    """Utility function to measure execution time using a loop."""
    start_time = time.perf_counter()
    for _ in range(iterations):
        func(*args)
    end_time = time.perf_counter()
    elapsed = end_time - start_time
    print(
        f"{name: <25} | Iterations: {iterations: <6} "
        f"| Total Time: {elapsed:.5f}s "
        f"| Avg Time: {(elapsed / iterations) * 1e6:.2f} µs")
```

## Problem 1
Given a list of strings, concatenate them into one large string.

```python
# Generate random sequences
my_sequences = generate_random_strings(10, 25)

# Version 1
result = ""
for s in my_sequences:
    result += s

# Version 2
result = "".join(my_sequences)
```

I created several tests, each for one of the Versions above and in two different conditions/scenerios. The above code was turned into functions in the code. I used benchmark which was mentioned previously to test the time it takes to complete all iterations and the avg. time for each.

```python
def p1_version1_loop(sequences):
    result = ""
    for s in sequences:
        result += s
    return result


def p1_version2_join(sequences):
    return "".join(sequences)


def run_problem_1_experiments():
    print("Problem 1: List Concatenation")

    # Scenario A: Many short strings
    seqs_many_short = generate_random_strings(1000, 10)
    print("\nScenario 1A: 1000 strings of length 10")
    benchmark("Version 1 (+= loop)", p1_version1_loop, seqs_many_short, iterations=1000)
    benchmark("Version 2 (join)", p1_version2_join, seqs_many_short, iterations=1000)

    # Scenario B: Fewer long strings
    seqs_few_long = generate_random_strings(10, 10000)
    print("\nScenario 1B: 10 strings of length 10000")
    benchmark("Version 1 (+= loop)", p1_version1_loop, seqs_few_long, iterations=1000)
    benchmark("Version 2 (join)", p1_version2_join, seqs_few_long, iterations=1000)
    print("\n")
```

Observations:
Version 2 ("".join()) is significantly faster than Version 1 (the += loop).
The performance gap widens as the number of strings in the list increases. For very large lists, the += loop becomes noticeably slower.
My Explanation: In Python, strings are immutable. This means they cannot be changed after they are created. 
Version 1 (+=): In a naive implementation, result += s creates a brand-new string object in memory, copies the old result into it, and then appends s. If you do this 1,000 times, you are constantly allocating new memory and copying increasingly larger strings. This leads to time complexity, a longer time to completion.
Version 2 (.join()): This method is highly optimized in C. Python first iterates through the list to calculate the exact total length of the final string. It then allocates memory exactly once and copies all substrings directly into their correct positions. This results in time complexity, making it vastly superior for lists. 
Important to Remember: The builtin modules/functions for python are almost always written in highly optimized C, which significantly outmatch the speed of native pyhon code.

## Problem 2
Given (only) two strings, concatenate them into one large string

```python
my_sequences = generate_random_strings(2, 25)

# Version 1 -> string 'addition'
result = my_sequences[0] + my_sequences[1]

# Version 2 -> format()
result = "{}{}".format(my_sequences[0], my_sequences[1])

# Version 3 -> f-strings
result = f"{my_sequences[0]}{my_sequences[1]}"
```

I turned the above-mentioned operations into functions and implemented them in my code. Then I created timetest for each of the 3 Versions.

```python
def p2_version1_add(seqs):
    return seqs[0] + seqs[1]


def p2_version2_format(seqs):
    return "{}{}".format(seqs[0], seqs[1])


def p2_version3_fstring(seqs):
    return f"{seqs[0]}{seqs[1]}"


def run_problem_2_experiments():
    print("Problem 2: Two-String Concatenation")

    # Scenario A: Short strings
    seqs_short = generate_random_strings(2, 25)
    print("\nScenario 2A: 2 strings of length 25")
    benchmark("Version 1 (+ operator)", p2_version1_add, seqs_short, iterations=100000)
    benchmark("Version 2 (.format)", p2_version2_format, seqs_short, iterations=100000)
    benchmark("Version 3 (f-string)", p2_version3_fstring, seqs_short, iterations=100000)

    # Scenario B: Long strings
    seqs_long = generate_random_strings(2, 50000)
    print("\nScenario 2B: 2 strings of length 50000")
    benchmark("Version 1 (+ operator)", p2_version1_add, seqs_long, iterations=100000)
    benchmark("Version 2 (.format)", p2_version2_format, seqs_long, iterations=100000)
    benchmark("Version 3 (f-string)", p2_version3_fstring, seqs_long, iterations=100000)
```

Observations:
Version 1 (+ operator) and Version 3 (f-strings) are the fastest, often performing almost identically.
Version 2 (.format()) is consistently the slowest of the three methods.

Explanation:
Version 1 (+): For exactly two strings, the + operator is straightforward and extremely fast. Python simply allocates a new string object sized to hold both and copies the two memory blocks over.
Version 2 (.format()): This is the slowest because of the overhead involved. Python has to call the string .format() method, parse the   "{}{}" string at runtime to locate placeholders, and process the args.
Version 3 (f-strings): F-strings were introduced in Python to be fast. They are evaluated at compile time. Python converts f"{a}{b}" directly into efficient C-level string building operations. There is no format string parsing at runtime, making it functionally as fast as the native + operator while often being more readable.
